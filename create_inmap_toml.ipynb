## Create an inmap run file (toml)

```python
import os

def inmap_process_template(template_path, emis_file_path, run_name, output_path):
    """Reads a template file, replaces placeholders with actual paths, and saves the modified file."""
    with open(template_path, 'r') as f:
        content = f.read()
    
    # Replace placeholders
    content = content.replace('{emis_file_path}', emis_file_path)
    content = content.replace('{run_name}', run_name)

    print("new run file will be save at : ", output_path)
    # Write the modified content to a new file
    with open(output_path, 'w') as f:
        f.write(content)

# Template setup
inmap_run_file_dir = '/Users/yunhalee/Documents/LOCAETA/RCM/INMAP/inmap-1.9.6-gridsplit/eval'
inmap_run_template = 'nei2020Config_template.toml'
run_template = os.path.join(inmap_run_file_dir, inmap_run_template)

# emission file path and run name
emis_dir = '/Users/yunhalee/Documents/LOCAETA/eGRID_emissions'
run_name_list = ["current_2020","decarb95_2050","highREcost_2050", "current_2020_base","decarb95_2050_base","highREcost_2050_base" ]


for run_name in run_name_list: 
     inmap_run_file_output = os.path.join(inmap_run_file_dir, f'nei2020Config_{run_name}.toml')
     emis_file_path = os.path.join(emis_dir, f'{run_name}.shp')

     print(f'{inmap_run_file_output} is created')
     inmap_process_template(run_template, emis_file_path, run_name, inmap_run_file_output)

```

## submit the inmap run job

Note that I am setting to do one run at a time.

```python
import os

run_name = "highREcost_2050_base" # "decarb95_2050" # "current_2020_base"
working_dir = "/Users/yunhalee/Documents/LOCAETA/RCM/INMAP/inmap-1.9.6-gridsplit"
config_file = f"eval/nei2020Config_{run_name}.toml"


#make output directory
os.makedirs(os.path.join(working_dir,'outputs', run_name), exist_ok=True)

# Escape all quotes for AppleScript
command = f'cd "{working_dir}"; caffeinate -i ./inmap run steady -s --config {config_file}'
escaped_command = command.replace('"', '\\"')

osa_command = f'''
osascript -e 'tell application "Terminal" to do script "{escaped_command}"'
'''

os.system(osa_command)

```
